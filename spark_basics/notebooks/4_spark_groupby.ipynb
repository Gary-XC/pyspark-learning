### PySpark GroupBy And Aggregate Functions

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('GroupBy And Aggregate').getOrCreate()
spark
```

```python
df_spark = spark.read.csv('spark_groupby_agg.csv', header=True, inferSchema=True)
df_spark.show()
```

```python
df_spark.printSchema()
```

#### GroupBy Operations


```python
df_spark.groupBy('Name')
```

```python
df_spark.groupBy('Name').sum().show()
# since there are some people with more than one salary type, grouping the names and then adding up all the values will result in the total salary of everyone
```

```python
# Grouping by departments to find which department has the highest salary

df_spark.groupBy('Departments').sum().show()
```

```python
# finding the max salary for each person (by grouping by Names: since people have more than one salary) and using the .max() function

df_spark.groupBy('Name').max().show()
```

```python
# the mean or average salary for an employee for each department

df_spark.groupBy('Departments').mean().show()
```

```python
# finding out the number of employees at each department: using count

df_spark.groupBy('Departments').count().show()
```

#### Aggregate Functions

```python
df_spark.agg({'Salary' : 'sum'}).show()
# the sum of everyone's salary
```
