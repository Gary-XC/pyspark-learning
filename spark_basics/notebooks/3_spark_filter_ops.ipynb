### Filler Operations
- Filter Operations
- &, |, ==
- ~

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('Filter Operations Practice').getOrCreate()
```

```python
df_spark = spark.read.csv('spark_filter_ops.csv', header=True, inferSchema=True)
df_spark.show()
```

#### Filter Operations
- similar to sql or pandas operations
    - df_spark_sample.filter("`column_filtered <= Value`"), or df_spark_sample.filter(df_spark_sample["column_filtered"] <= Value")
        - <= can be any comparison operators: !=, ==, <, >, <=, >=
    - df_spark_sample.filter("").select(['column_selected_1', 'column_selected_2'])

```python
# People making a salary less than or equal to 20,000

df_spark.filter("Salary <= 20000").show()
```

```python
df_spark.filter(df_spark["Salary"] <= 20000).show()
# this is another method of getting the same result, and should be the default and first method to be used when filtering, since this is the only way that works when multiple conditions are considered
```

```python
# selecting only certain columns and information using the select function

df_spark.filter("Salary <= 20000").select(['Age', 'Name']).show() # since only age and name are chosen, they are the only values in the resultant table
```

```python
# Using multiple conditions: are grouped using () brackets, and are separated by and (&) / or (|) comparators
# when using multiple conditions, only the second method(where the pyspark column is called as well) works

df_spark.filter((df_spark["Salary"] <= 20000) & (df_spark["Experience"] >= 5)).show()
```

Not Condition: ~
- is the inverse operation


```python
df_spark.filter(~(df_spark["Salary"] <= 20000)).show()
# shows everyone making more than 20,000
```
