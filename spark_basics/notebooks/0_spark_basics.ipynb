##### Methods and Functions Similar To Pandas And Numpy

```python
import pandas as pd
import numpy as np

df_pandas = pd.read_csv('spark_basics.csv')
df_pandas.head(), type(df_pandas)
```

```python
df_pyspark = spark.read.csv('spark_basics.csv')
# this method has its own built in headers for each column in the dataset

df_pyspark.show() # so if the first value in each column is the header name, it gets added to the columns total values
# as seen in the example below, the headers for each column (name, age) are considered values under the new columns that spark made: _c0, _c1
```

```python
df_pyspark = spark.read.option("header", "true").csv("spark_basics.csv")
# spark read has an option, where you can choose to use headers, if there are headers included in the dataset
# in this example the headers are now populating in the dataframe, and there are no default headers provided by pyspark

df_pyspark.show()
```
